```python
from pathlib import Path
import subprocess, sys, os

os.chdir(f"/content/{REPO}")

def sh(*cmd):
    r = subprocess.run(cmd, capture_output=True, text=True)
    print((r.stdout + r.stderr).replace(TOKEN, "***").strip())
    return r.returncode

def sync(msg):
    sh("git", "add", "-A")
    sh("git", "commit", "-m", msg)
    sh("git", "push", "origin", "main")

# structure
for d in ["notebooks", "src/lrlab", "tests", "reports/figures", "data"]:
    Path(d).mkdir(parents=True, exist_ok=True)
for d in ["reports/figures", "data"]:
    Path(d, ".gitkeep").touch()

Path("requirements.txt").write_text(
    "numpy\npandas\nmatplotlib\nseaborn\nscikit-learn\nscipy\npytest\n")

Path("pyproject.toml").write_text('''[build-system]
requires = ["setuptools>=68"]
build-backend = "setuptools.build_meta"

[project]
name = "lrlab"
version = "0.1.0"
requires-python = ">=3.10"

[tool.setuptools.packages.find]
where = ["src"]
''')

Path("src/lrlab/__init__.py").write_text('"""Linear regression lab: from-scratch implementations."""\n')
Path("src/lrlab/config.py").write_text('"""Global configuration."""\nSEED = 42\nFIG_DIR = "reports/figures"\n')
Path("tests/test_smoke.py").write_text(
    "from lrlab.config import SEED\n\n\ndef test_seed_is_fixed():\n    assert SEED == 42\n")

# install package in editable mode + run tests
sh(sys.executable, "-m", "pip", "install", "-q", "-r", "requirements.txt")
sh(sys.executable, "-m", "pip", "install", "-q", "-e", ".")
sh(sys.executable, "-m", "pytest", "-q")

sync("chore: scaffold project structure and smoke test")
sh("git", "log", "--oneline")
```
